```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread(r"C:\Users\Asus\OneDrive\Desktop\Computer_vision\images\giraffe-Kenya.png")
```

```python
image_gray = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)

equalized_image = cv2.equalizeHist(image_gray)

clahe = cv2.createCLAHE(clipLimit=2.0,tileGridSize=(4,4))

clahe_equalized_image = clahe.apply(image_gray)

plt.figure(figsize=(12,8))
plt.subplot(1,3,1)
plt.imshow(image_gray,cmap='gray')
plt.title("Original Grayscale Image")

plt.subplot(1,3,2)
plt.imshow(equalized_image,cmap='gray')
plt.title("Histogram Equalized Image")

plt.subplot(1,3,3)
plt.imshow(clahe_equalized_image,cmap='gray')
plt.title("CLAHE Equalized Image")

plt.tight_layout()
plt.show()
```

```python
image_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
# YUV 

yuv_image = cv2.cvtColor(image,cv2.COLOR_BGR2YUV)

yuv_image[:,:,0] = clahe.apply(yuv_image[:,:,0])

yuv_image_BGR = cv2.cvtColor(yuv_image,cv2.COLOR_YUV2BGR)

yuv_image_RGB = cv2.cvtColor(yuv_image_BGR,cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12,8))
plt.subplot(1,3,1)
plt.imshow(image_rgb)
plt.title("Original RGB Image")


plt.subplot(1,3,2)
plt.imshow(yuv_image_RGB)
plt.title(" CLAHE Equalized RGB Image")

plt.tight_layout()
plt.show()


```
