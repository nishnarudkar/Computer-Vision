```python
import torch
```

```python
tensor1 = torch.tensor([1,2,3])
tensor2 = torch.tensor([4,5,6])

print(tensor1.ndim)
print(tensor2.ndim)
```

```python
stacked_tesnor_dim0 = torch.stack((tensor1, tensor2), dim = 0)
print(" Shape : ", stacked_tesnor_dim0.shape)
print(stacked_tesnor_dim0)


stacked_tensor_dim1 = torch.stack((tensor1, tensor2), dim = 1)
print(" Shape : ", stacked_tensor_dim1.shape)
print(stacked_tensor_dim1)
```

```python
tensor_3 = torch.tensor([7,8,9])

stacked_tensor_dim_0_new = torch.stack((tensor1, tensor2, tensor_3), dim = 0)
print(" Shape : ", stacked_tensor_dim_0_new.shape)
print(stacked_tensor_dim_0_new)
```

```python
# Stacking 2D tensors

tensor1 = torch.tensor([[1,2,3],[4,5,6]])
tensor2 = torch.tensor([[7,8,9],[10,11,12]])


stacked_tensor_2d = torch.stack((tensor1, tensor2), dim = 0)
print(" Shape : ", stacked_tensor_2d.shape)
print(stacked_tensor_2d)
```

```python
stacked_tensor_2d = torch.stack((tensor1, tensor2), dim = 1)
print(" Shape : ", stacked_tensor_2d.shape)
print(stacked_tensor_2d)
```

```python
stacked_tensor_2d = torch.stack((tensor1, tensor2), dim = 2)
print(" Shape : ", stacked_tensor_2d.shape)
print(stacked_tensor_2d)
```

```python
# Concatenation

concatenated_tensor = torch.cat((tensor1, tensor2), dim = 0)
print(" Shape : ", concatenated_tensor.shape)
print("Dimension: ", concatenated_tensor.ndim)
print(concatenated_tensor)
```

```python
# Concatenation

concatenated_tensor = torch.cat((tensor1, tensor2), dim = 1)
print(" Shape : ", concatenated_tensor.shape)
print("Dimension: ", concatenated_tensor.ndim)
print(concatenated_tensor)
```
