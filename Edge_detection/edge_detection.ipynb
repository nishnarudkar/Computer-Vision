```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread(r"C:\Users\Asus\OneDrive\Desktop\Computer_vision\images\giraffe-Kenya.png", cv2.IMREAD_GRAYSCALE)
```

```python
# Sobel Edge Detection

sobel_x = cv2.Sobel(image,cv2.CV_64F,1,0,ksize=3)
sobel_y = cv2.Sobel(image,cv2.CV_64F,0,1,ksize=3)
sobel_combined = cv2.magnitude(sobel_x, sobel_y)
```

```python
# Laplacian Edge Detection

laplacian = cv2.Laplacian(image, cv2.CV_64F, ksize=3)
laplacian_abs = cv2.convertScaleAbs(laplacian)
```

```python
# Canny edge detection

canny_edges = cv2.Canny(image, 100, 200)

titles = ["Original Image", "Sobel Edge Detection", "Laplacian_all","Laplacian_abs", "Canny Edge Detection"]
images = [image, sobel_combined, laplacian, laplacian_abs , canny_edges]

plt.figure(figsize= (20,16))
for i in range(len(titles)):
    plt.subplot(3, 2, i + 1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')
plt.show()
```
