```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("C:\\Users\\Asus\\OneDrive\\Desktop\\Computer_vision\\images\\butterfly.jpg")
```

```python
image_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

# Flip

```python
# 1 ----> horizontal flip
# 0 ----> vertical flip
# -1 ----> both horizontal and vertical flip

flip_horizontally = cv2.flip(image, 1)
plt.imshow(flip_horizontally)
```

```python
flip_vertically = cv2.flip(image, 0)
plt.imshow(flip_vertically)
```

```python
flip_both = cv2.flip(image, -1)
plt.imshow(flip_both)
```

```python
titles = ["Original Image","Flip Horizontally","Flip Vertically","Flip Both"]
images = [image_rgb,
          cv2.cvtColor(flip_horizontally, cv2.COLOR_BGR2RGB),
          cv2.cvtColor(flip_vertically, cv2.COLOR_BGR2RGB),
          cv2.cvtColor(flip_both, cv2.COLOR_BGR2RGB)
          ]
plt.figure(figsize=(12,8))
for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis('off')
plt.tight_layout()
plt.show()
```

# Rotation

```python
rotate_90_clockwise = cv2.rotate(image_rgb, cv2.ROTATE_90_CLOCKWISE)
rotate_90_counterclockwise = cv2.rotate(image_rgb, cv2.ROTATE_90_COUNTERCLOCKWISE)
rotate_180 = cv2.rotate(image_rgb, cv2.ROTATE_180)

titles = ["Original Image","90 degree Clockwise","90 degree Counterclockwise","180 degrees"]
images = [image_rgb, rotate_90_clockwise, rotate_90_counterclockwise, rotate_180]
plt.figure(figsize=(12,8))
for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis('off')
```
