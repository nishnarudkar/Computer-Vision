```python
# Accessing elements in a tensor

'''
tensor[row,column]
'''


import torch
```

```python
tensor = torch.tensor([[10,20,30],[40,50,60],[70,80,90]])
print(tensor.ndim)
print(tensor.shape)
```

```python
print("Element at (0,1): ", tensor[0,1])
```

```python
# Access first row
tensor[0]
```

```python
# Acess first column
tensor[:,0]
```

```python
### Slicing


print("First two rows: ", tensor[:2])
```

```python
print("First two columns: ", tensor[:,:2])
```

```python
# Middle elements (row 1 , column 1 and 2 only)
print(tensor[1,1:3])
```

```python
# Fancy Indexing
# (0,1) --> 20
# (2,2) --> 90
selected_elements = tensor[[0,2],[1,2]]
print(selected_elements)
```

```python
# Boolean Indexing

mask = tensor > 50
print(mask)
print(tensor[mask])
```

```python
# Changing tensor values via indexing

print(tensor)
tensor[0,1] = 25
print(tensor)
```

```python
tensor[:,0] = torch.tensor([100,200,300])
print(tensor)
```

```python
indices = torch.tensor([0,2])
selected_rows = torch.index_select(tensor, dim=0, index=indices)
print(selected_rows)
```

```python
# Using tensor with steps

print(tensor[:,::2])
```

```python

torch.flip(tensor, dims=(0,))
```
