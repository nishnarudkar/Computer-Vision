```python
import torch
```

```python
random_tensor= torch.rand(2,5,3)
print(random_tensor)
print(random_tensor.ndim)
print(random_tensor.shape)
```

```python
torch.manual_seed(42)

random_tensor_1= torch.rand(2,5,3)
random_tensor_2= torch.rand(2,5,3)

print(random_tensor_1)
print(random_tensor_2)
```

```python
## Generate Random Integers

# torch.randint(low,high, size)

random_integers = torch.randint(-10,10,(5,5))
print(random_integers)
```

```python
# Generate random numbers from normal distribution

# torch.normal(mean,std,size)

random_numbers_normal = torch.normal(0.5,2,(5,))
print(random_numbers_normal)
```

```python
## Create noise image using Pytorch

import matplotlib.pyplot as plt

# h,w of image 256 x 256

uniform_noise = torch.rand((256,256))
plt.imshow(uniform_noise, cmap = "grey")
plt.title("Uniform noise image")
plt.axis("off")
plt.show()
```

```python
## Create noise image using Pytorch

import matplotlib.pyplot as plt

# h,w of image 256 x 256

uniform_noise_rgb = torch.rand((256,256,3))
plt.imshow(uniform_noise_rgb.numpy())
plt.title("Uniform noise image rgb")
plt.axis("off")
plt.show()
```

```python
print("Shape",uniform_noise_rgb.shape)
print("Dimension",uniform_noise_rgb.ndim)
print("Min value", torch.min(uniform_noise_rgb))
print("Max value", torch.max(uniform_noise_rgb))
```
