```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread(r"C:\\Users\Asus\\OneDrive\Desktop\\Computer_vision\\images\\giraffe-Kenya.png")
```

```python
image_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
kernel_s = 5

average_blur = cv2.blur(image_gray, (kernel_s,kernel_s))

plt.imshow(average_blur, cmap = "gray")
```

```python
kernel_s = 5
gaussian_blur= cv2.GaussianBlur(image_gray, (kernel_s, kernel_s),0)
plt.imshow(gaussian_blur, cmap = "gray")
```

```python
median_blur = cv2.medianBlur(image_gray,11)
plt.imshow(median_blur, cmap = "gray")
```

```python
plt.figure(figsize= (14,12))

plt.subplot(2,2,1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(average_blur, cmap = "gray")
plt.title("Average Blur")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(gaussian_blur, cmap = "gray")
plt.title("Gaussian Blur")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(median_blur, cmap = "gray")
plt.title("Median Blur")
plt.axis("off")
```
