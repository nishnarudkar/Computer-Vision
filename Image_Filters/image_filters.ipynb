```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread(r"C:\Users\Asus\OneDrive\Desktop\Computer_vision\images\filter.jpg", cv2.IMREAD_GRAYSCALE)
plt.imshow(image, cmap='gray')
```

```python
x = -20
horizontal_kernel = np.array([[-1,-1,-1],[x,x,x],[1,1,1]])

horizontal_filtered_image = cv2.filter2D(image, -1, horizontal_kernel)
plt.imshow(horizontal_filtered_image, cmap = "gray")
```

```python
vertical_kernel = np.array([[-1,x,1],[-1,x,1],[-1,x,1]])
vertical_filtered_image = cv2.filter2D(image, -1, vertical_kernel)
plt.imshow(vertical_filtered_image, cmap = "gray")
```

```python
diagonal_kernel = np.array([[x,-1,-1],
                            [1,x,-1],
                            [1,1,x]])
diagonal_filtered_image = cv2.filter2D(image, -1, diagonal_kernel)
plt.imshow(diagonal_filtered_image, cmap = "gray")
```

```python
plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
plt.title("Original Image")
plt.imshow(image, cmap = "gray")

plt.subplot(2,2,2)
plt.title("Horizontal Filtered Image")
plt.imshow(horizontal_filtered_image, cmap = "gray")

plt.subplot(2,2,3)
plt.title("Vertical Filtered Image")
plt.imshow(vertical_filtered_image, cmap = "gray")

plt.subplot(2,2,4)
plt.title("Diagonal Filtered Image")
plt.imshow(diagonal_filtered_image, cmap = "gray")
```
